```python
# ==========================================
# 1. Necessary Libraries Install & Import
# ==========================================
import numpy as np
import plotly.graph_objects as go

# ==========================================
# 2. Elliptic Curve & Lattice Parameters (38970.a1)
# ==========================================
omega_1 = 0.95296796
im_tau = 1.63721906
omega_2_im = omega_1 * im_tau  # approx 1.5602

# Normalized flat-torus coordinates for P=(-8,11) and Q=(-7,18)
# Periodic boundaries: [0, 2pi)
u_P = np.array([0.42 * 2 * np.pi, 0.28 * 2 * np.pi])
u_Q = np.array([0.65 * 2 * np.pi, 0.74 * 2 * np.pi])

# ==========================================
# 3. 3D Torus Geometric Embedding
# ==========================================
# Major radius R (distance from donut hole to tube center)
# Minor radius r (tube thickness)
R, r = 3.0, 1.2

grid_res = 60
theta = np.linspace(0, 2 * np.pi, grid_res)  # Poloidal angle (along omega_1)
phi = np.linspace(0, 2 * np.pi, grid_res)    # Toroidal angle (along omega_2)
THETA, PHI = np.meshgrid(theta, phi)

# Embedding (u1, u2) -> R^3
X = (R + r * np.cos(THETA)) * np.cos(PHI)
Y = (R + r * np.cos(THETA)) * np.sin(PHI)
Z = r * np.sin(THETA)

# ==========================================
# 4. Stream Function & Helicity Density Field
# ==========================================
# Periodic distance approximation on torus
def torus_dist_sq(t, p, target):
    dt = np.minimum(np.abs(t - target[0]), 2*np.pi - np.abs(t - target[0]))
    dp = np.minimum(np.abs(p - target[1]), 2*np.pi - np.abs(p - target[1]))
    return dt**2 + dp**2

# Green function proxy for vortex points P and Q (core size epsilon)
eps2 = 0.05
psi_P = -np.log(torus_dist_sq(THETA, PHI, u_P) + eps2)
psi_Q = -np.log(torus_dist_sq(THETA, PHI, u_Q) + eps2)

# Helicity density / Cross interaction energy proxy
# h(x) ~ grad(psi_P) . curl(grad(psi_Q))
H_density = psi_P * psi_Q

# ==========================================
# 5. Coordinates of Generators P and Q in 3D
# ==========================================
P_3D = [
    (R + r * np.cos(u_P[0])) * np.cos(u_P[1]),
    (R + r * np.cos(u_P[0])) * np.sin(u_P[1]),
    r * np.sin(u_P[0])
]

Q_3D = [
    (R + r * np.cos(u_Q[0])) * np.cos(u_Q[1]),
    (R + r * np.cos(u_Q[0])) * np.sin(u_Q[1]),
    r * np.sin(u_Q[0])
]

# ==========================================
# 6. Interactive 3D Plot Render
# ==========================================
fig = go.Figure()

# Surface: Torus with Helicity Density Heatmap
fig.add_trace(go.Surface(
    x=X, y=Y, z=Z,
    surfacecolor=H_density,
    colorscale='Plasma',
    opacity=0.88,
    colorbar=dict(title='Helicity Density h(x)'),
    showscale=True
))

# Vortex Point P marker
fig.add_trace(go.Scatter3d(
    x=[P_3D[0]], y=[P_3D[1]], z=[P_3D[2]],
    mode='markers+text',
    marker=dict(size=8, color='cyan', symbol='circle'),
    text=['<b>P (-8, 11)</b>'],
    textposition='top center',
    name='Generator P'
))

# Vortex Point Q marker
fig.add_trace(go.Scatter3d(
    x=[Q_3D[0]], y=[Q_3D[1]], z=[Q_3D[2]],
    mode='markers+text',
    marker=dict(size=8, color='lime', symbol='diamond'),
    text=['<b>Q (-7, 18)</b>'],
    textposition='top center',
    name='Generator Q'
))

# Layout Configuration for Mobile Friendly Touch
fig.update_layout(
    title='3D Topological Fluid Helicity on Torus (Curve 38970.a1)',
    scene=dict(
        xaxis=dict(showgrid=False, zeroline=False, showticklabels=False),
        yaxis=dict(showgrid=False, zeroline=False, showticklabels=False),
        zaxis=dict(showgrid=False, zeroline=False, showticklabels=False),
        aspectmode='data'
    ),
    margin=dict(l=0, r=0, b=0, t=40)
)

fig.show()
```
